```python
## Add item
# append: adds item in the end

l1 = ["apple","banana","mango"]
l1.append("pineapple")
l1
```

```python
## Add item
# insert: adds item in the certain position

l2 = ["apple","banana","mango"]
l2.insert(2,"watermelon")
l2
```

```python
## Combine both lists

list1 = [1,2,3,4]
list2 = [90.32,45]
list3 = list1 + list2
list3
```

```python
## Combining both lists using append

list4 = [1,2,3,4]
list5 = [90,32,45]
list4.append(list5)
list4
```

```python
## REMOVE
# remove()

l3 = ['apple', 'banana', 'watermelon', 'mango']
l3.remove("apple")
l3
```

```python
# pop(): removes last element by default

l3 = ['apple', 'banana', 'watermelon', 'mango']
l3.pop()
l3
```

```python
# pop(index): removes desired element using provided index

l3 = ['apple', 'banana', 'watermelon', 'mango']
l3.pop(0)
l3
```

```python
# delete by index: delete element by index using del

l3 = ['apple', 'banana', 'watermelon', 'mango']
del l3[1]
l3
```

```python
## clear(): empty the list

l4 = [23,45,24,35,41]
l4.clear()
l4
```

```python
## dl list_name: deletes the complete list

l4 = [23,45,24,35,41]
del l4
l4
```

```python
## COPY

# shallow copy: creates another list by copying the input list

l1 = [23,24,25,26]
l2 = l1.copy()
l2
```

```python
# Assignement/Reference: Both list refers to the same memory location

l3 = [23,24,25,26]
l4 = l3
l4
```

```python
## Shallow copy
l1 = [23,24,25,26]
l2 = l1.copy()
print("SHALLOW COPY:")
print("Initial values")
print(l1)
print(l2)

print("Values after adding elements")
l1.append(27)
print(l1)
l2.append(45)
print(l2)

print("\n")

## Assignment operator
l3 = [90,56,10,40]
l4 = l3
l4
print("ASSIGNMENT OPERATOR: ")
print("Initial values")
print(l3)
print(l4)

print("Values after adding elements")
l3.append(29)
print(l3)
l4.append(89)
print(l4)
```

```python
## Joining / Merging

l1 = [22,33,44]
l2 = [66,77,88]
l1.extend(l2)
l1
```

```python
## SORT

# sort(): sort in place
l1 = [22,33,44]
l1.sort()
l1
```

```python
# sorted(): creates a sortd copy

l1 = [22,33,44]
l2 = sorted(l1)
l2
```

```python
# sort(revrse = True): sort reverse

l1 = [22,33,44]
l1.sort(reverse = True)
l1
```

```python
# reverse(): reverse the list

l1 = [22,33,44]
l1.reverse()
l1
```

```python
# REVERSE BY SLICING

l1 = [12,23,34,45]
l1[::-1]
```

```python
## LENGTH

l1 = [22,33,44,66,55]
print(len(l1))
```

```python
## SEARCH

l1 = [22,33,44,66,55]
l1.index(33)
```

```python
## COUNT

l1 = [22,33,44,66,55,22]
l1.count(22)
```

```python
## in operator

l1 = [22,33,44,66,55,22]
33 in l1
```

```python
## for loop

l1 = [22,33,44,66,55,22]
for i in l1:
    print(i, l1.index(i))
```
