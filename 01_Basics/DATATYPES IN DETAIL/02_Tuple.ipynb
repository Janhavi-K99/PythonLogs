```python
### TUPLE

# Example Tuple: myTuple = ("apple", "banana", "mango")
# Note: Tuples are ordered and immutable. Once created, tuple items cannot be directly added, removed, or changed.
```

```python
## Access
t1 = ("mango", "banana", "apple")
t1[-1]
```

```python
## slicing

t1 = ("mango", "banana", "apple")
t1[-3:-1]
```

```python
## reverse slicing

t1 = ("mango", "banana", "apple")
t1[::-1]
```

```python
## add/join
# Joins two tuples and creates a new tuple

tuple1 = ("apple", "banana")
tuple2 = ("mango", "orange")
t3 = tuple1+tuple2
t3
```

```python
## del
# we can delete but cannot clear the tuple

tuple1 = ("apple", "banana")
del tuple1
tuple1
```

```python
tuple1 = ("apple", "banana","orange","mango")
```

```python
# check in
tuple1 = ("apple", "banana","orange","mango")
"apple" in tuple1
```

```python
# count
tuple1 = ("apple", "banana","orange","mango")
tuple1.count("mango")
```

```python
# length()
len(tuple1)
```

```python
# assignment
# Assigns the tuple to another variable. Since tuples are immutable, a separate copy method is generally not required
t01 = (1,2,3,4,5)
t02 = t01
t02
```

```python
# rpeat
# Repeats the tuple the specified number of times. Returns in the form of list

t03 = (1,2)
t03*4
```

```python
# sort

# as index changing is also not allowed sort is also not applicable
```
