```python
### SET
## Example Set: mySet = {"apple", "banana", "mango"}
## Note: Sets are unordered and do not allow duplicate values.
```

```python
## ADD

s1 = {"apple", "banana", "mango", "orange"}
s1.add("muskmelon")
s1
```

```python
## UPDATE

s1 = {"apple", "banana", "mango", "orange"}
abc = {"watermelon","grapes"}
s1.update(abc)
s1
```

```python
## REMOVE
# Description: Removes the specified item. Gives an error if the item does not exist.
s1 = {'apple', 'banana', 'grapes', 'mango', 'orange', 'watermelon'}
s1.remove("orange")
s1
```

```python
## discard()
# Description: Removes the specified item. Does not give an error if the item does not exist.
s1.discard("orange")
s1
```

```python
# pop()
# Description: Removes and returns an arbitrary item from the set.
s1 = {'apple', 'banana', 'grapes', 'mango', 'orange', 'watermelon'}
s1.pop()
```

```python
# clear()
# Description: Removes all items from the set.
s1 = {'apple', 'banana', 'grapes', 'mango', 'orange', 'watermelon'}
s1.clear()
s1
```

```python
s1 = {'apple', 'banana', 'grapes', 'mango', 'orange', 'watermelon'}
del s1
s1
```

```python
## update(): combine sets

set1 = {1, 2, 3}
set2 = {3, 4, 5}
set1.update(set2)
print(set1)
print(set2)
```

```python
## union(): combine sets

set1 = {1, 2, 3}
set2 = {3, 4, 5}

new_set = set1.union(set2)
print(new_set)
print(set1)
print(set2)
```

```python
# difference bet update and union
# set1.union(set2) -> crates new set with all elements combines & original sets remian as it is
# set1.update(set2) -> updates set1 by combining all the elements from the set2. st2 remian as it is
```
