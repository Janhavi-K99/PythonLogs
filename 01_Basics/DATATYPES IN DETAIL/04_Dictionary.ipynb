```python
## Adding single element

d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d1['pin'] = 444005
d1
```

```python
## Adding multiple elements

d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d1.update({'pin' : 444005, 'area': 'katraj'})
d1
```

```python
## clear():

d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d1.clear()
d1
```

```python
## del

d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
del d1
d1
```

```python
## deep copy
# creates new dict & stores at new memory loc
d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d1.copy()
```

```python
## shallow copy
# refers to same memory loc
d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d2 = d1
d2
```

```python
## Merging

d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d2 = {'pin': 444005, 'area': 'katraj'}
newDict = dict(d1,**d2)
newDict
```

```python
d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune'}
d2 = {'pin': 444005, 'area': 'katraj'}
newDict = {**d1,**d2}
newDict
```

```python
## reverse
# creates and stores at new memory location
d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune', 'pin': 444005, 'area': 'katraj'}
dict(reversed(list(d1.items())))
```

```python
## sorting
# creates and stores at new memory location
d1 = {'name': 'Rajesh', 'age': 20, 'city': 'Pune', 'pin': 444005, 'area': 'katraj'}
dict(sorted(list(d1.items())))
```
