```python
l1 = []
for i in range(0,15):
    if i<5:
        l1.append(i*2)
    elif i<10:
        l1.append(i/2)
    else:
        l1.append(i+2)  
l1
        
```

```python
# [exp1 if_cond else exp2 if_cond else exp3 for i in range(15)]

[i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)]
```

```python
## Dictionary Comprehension
# {exp:exp_2 for loop}
# {key:value for loop}

{i:i*2 if i<5 else i/2 if i<10 else i+2 for i in range(15)}
```

```python
l1 = [30,25,34,45,90,20,29,98,93,94,91,12,15,62,-9,105]
l2 = []
for i in l1:
    if i<27:
        l2.append("FAIL")
    elif i>=27 and i<=100:
        l2.append("PASS")
    else:
        l2.append("INVALID")  
l2
```

```python
list1 = [30,25,34,45,100,90,20,29,98,93,94,91,12,15,62,-1,500]

["PASS" if i>=27 and i<=100 else "FAIL" if i<27 else "INVALID" for i in list1]
```
