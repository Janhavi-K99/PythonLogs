```python
### RETURN TYPE

def add_Values(x,y):
    z = x+y
    return z

result = add_Values(3,9)
result
```

```python
def add_Values(x,y):
    z = x+y
    return z

result = add_Values(3,9)

def multiply_Values(a,b):
    c = a*b
    return c

result02 = multiply_Values(result,100)
print(result02)
```

```python
### NON RETURN TYPE

def add_Values(x,y):
    z = x+y
    print(z)

result = add_Values(3,9)
print(result)
```

```python
### Use of yield keyword

def newFunc(x):
    for i in range(1,x):
        yield i

result01 = newFunc(10)

print(next(result01))
print(next(result01))
print(next(result01))
```

```python
# This same cannot be achieved by return. return keyword means terminating the program. If tried to do so eveytime the previous memory is erased & the function is called from the start again and again.

def newFunc(x):
    for i in range(1,x):
        return i

result01 = newFunc(10)

print(result01)
print(result01)
print(result01)
```
