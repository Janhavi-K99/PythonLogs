```python
### EXCEPTION HANDLING
## try block: lets you test a block of code for errors.
## except block: lets you handle the error.
## finally block: lets you execute code, regardless of the result of the try and except block.
## else keyword: to define a block of code to be executed if no errors were raised
```

```python
a = 5
b = 0

try:
    a/b
    #print(b/a)
except: 
    print("Something went wrong")
```

```python
a = 5
b = 0

try:
    a/b  
except: 
    c = b/a
    print(c)
```

```python
a = 5
b = 1

try:
    c = a/b
    print(c)  
except: 
    print("Something went wrong")
finally:
    print("Program closed")
```

```python
a = 5
b = 0

try:
    c = a/b
    print(c)   
except: 
    print("Something went wrong")
finally:
    print("Program closed")
```
