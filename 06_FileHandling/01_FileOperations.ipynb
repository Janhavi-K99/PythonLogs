```python
## CREATE & OPEN FILE
path = "file.txt"
file = open(path,'w')
file.close()
```

```python
## WRITE
path = "file_01.txt"
file = open(path,'w')
file.write("this is python class")
file.close()
```

```python
path = "file_01.txt"
file = open(path,'w')
file.writelines("Hello everyone" "\n"
                "this is python class" "\n"
                "lets revise")
file.close()
```

```python
## READ
path = "file_01.txt"
file = open(path,'r')
readOP = file.read()
print(readOP)
file.close()
```

```python
path = "file_01.txt"
file = open(path,'r')
readOP = file.readline()
print(readOP)
file.close()
```

```python
path = "file_01.txt"
file = open(path,'r')
readOP = file.readlines()
print(readOP)
file.close()
```

```python
path = "file.txt"
file = open(path,'w')
file.write("1) Today is wednesday ")
file.write("2) lets go to class")

file = open(path,'r')
readOP = file.readlines()
for i in readOP:
    print(i)
file.close()
```

```python
## with

path = "file_03.txt"
with open(path,'w') as file:
    file.write("Today is wednesday")
    
with open(path,'r') as file:
    print(file.read())
```
