```python
# Home work
# Find the maching values with the index Number
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]
col_03 = ["Black","LightGreen","White" ,"Pink"]

for i in col_01:
    for j in col_02:
        for k in col_03:
            if (i == j == k):
                print("Matched")
                print("col_01: index of", i, col_01.index(i))
                print("col_02: index of", i, col_02.index(j))
                print("col_03: index of", i, col_03.index(k))
```

```python
col_01 = ["Red", "Yellow", "LightGreen"]
col_02 = ["Black","White" ,"LightGreen"]
col_03 = ["Black","LightGreen","White" ,"Pink"]

def findMatch(col_01,col_02,col_03):
    for i in col_01:
        for j in col_02:
            for k in col_03:
                if (i == j == k):
                    return (i, col_01.index(i), col_02.index(j), col_03.index(k))
                
result = findMatch(col_01,col_02,col_03)
result
                    
```
