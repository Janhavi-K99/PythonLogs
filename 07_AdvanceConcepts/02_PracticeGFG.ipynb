```python
## Print Right Half Pyramid Star Pattern

# Input: N = 3
# Output: 
# *
# **
# ***

# Input: N = 5
# Output: 
# *
# **
# ***
# ****
# *****

def patternPrinting(n):
    for i in range(1,n):
        print("*"*i, end=" ")
        print()

patternPrinting(5)
```

```python
## Print Inverted Right Half Pyramid Pattern 

# Input: n = 5
# Output:
# *****
# ****
# ***
# **
# *

# Input: n = 3
# Output:
# ***
# **
# *


def patternPrinting(n):
    for i in range(n,0,-1):
        print("*"*i, end=" ")
        print()

patternPrinting(5)
```

```python
# Inverted Left Half Pyramid Star Pattern

# Input: N = 3
# Output: 
# ***
#   **
#     *

# Input: N = 5
# Output: 
# *****
#   ****
#     ***
#       **
#         *


def patternPrinting(n):
    for i in range(n,0,-1):
        print(" "*(n-i)+"*"*i, end=" ")
        print()

patternPrinting(5)
```

```python
# print half Diamond star pattern

# Input: N = 3
# Output:
# *
# **
# ***
# **
# *

# Input: N = 6
# Output:
# *
# **
# ***
# ****
# *****
# ******
# *****
# ****
# ***
# **
# *


def patternPrinting(n):
    for i in range(1,n):
        print("*"*i, end = " ")
        print()
    for j in range(n,0,-1):
        print("*"*j, end = " ")
        print()
patternPrinting(5)
```

```python
# Print the Diamond Shape

#     * 
#    * * 
#   * * * 
#  * * * * 
# * * * * * 
# * * * * * 
#  * * * * 
#   * * * 
#    * * 
#     * 

def patternPrinting(n):
    for i in range(1,n):
        print(" "*(n-i) + "* "*i, end=" ")
        print()
    for j in range(n,0,-1):
        print(" "*(n-j) + "* "*j, end=" ")
        print()
patternPrinting(5)
```
