ACCESS MODIFIERS METHODS

```python
class Student():
    def display_public(self):
        print("public method")
        
    def _display_protected(self):
        print("protected method")
        
    def __display_private(self):
        print("private method")
        
obj1 = Student()
obj1.display_public()
obj1._display_protected()
obj1.__display_private()
```

```python
class Student():
    def display_public(self):
        print("public method")
        
    def _display_protected(self):
        print("protected method")
        
    def __display_private(self):
        print("private method")
        
    def display(self):
        self.display_public()
        self._display_protected()
        self.__display_private()
        
obj2 = Student()
obj2.display()
        
```
