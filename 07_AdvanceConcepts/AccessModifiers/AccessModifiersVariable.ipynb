ACCESS MODIFIERS VARIABLES

```python
class Student:
    def __init__(self):
        self.name = "Rajesh"
        self._age = 30
        self.__company = "google"
        
    def display(self):
        print(self.name)
        print(self._age)
        print(self.__company)
    
obj1 = Student()
obj1.display()
```
