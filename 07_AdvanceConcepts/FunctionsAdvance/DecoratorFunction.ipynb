```python
def myFucntion(func):
    def wrapper():
        print("Before function is called.")
        func()
        print("After function is called")
    return wrapper

@myFucntion
def say_Hello():
    print("Hello!!")
    
say_Hello()
```

```python
def myFunc4(func):
    def wrapper(x,y):
        if type(x) or type(y) == str:
            x, y= int(x), int(y)
        func(x,y)
    return wrapper

@myFunc4
def division(x,y):
    print(x/y)
    
division("46", 4) 
```

```python
def myFunc4(func):
    def wrapper(x,y):
        if isinstance(x,str):
            x = int(x)
        if isinstance(y,str):
            y = int(y)
        if y>x:
            x,y = y,x
        func(x,y)
    return wrapper

@myFunc4
def division(x,y):
    print(x/y)
    
division("45", 90) 
```

```python
list01 = [1,2,3,4]
list02 = []

def myFunc5(func):
    def wrapper(l1):
        l1 = [int(i) if isinstance(i, str) else i for i in l1]
        l2 = func(l1)

        d1 = {}
        for i in range(len(l2)):
            d1[l1[i]] = l2[i]
        print(d1)
            
    return wrapper

@myFunc5
def originalFunc(l1):
    for i in l1:
        list02.append(i**2)
    return list02
        
originalFunc(list01)
```
