```python
### INSTANCE VARIABLES AND INSTANCE METHODS
```

```python
class Dog():
    def __init__(self):
        self.dogAge = 25
        
obj1 = Dog()
obj1.dogAge
```

```python
class Dog():
    def __init__(self,age):
        self.dogAge = age
        
    def func1(self):
        print(self.dogAge)
        
obj1 = Dog(14)
obj1.func1()
```

```python
class Dog():
    def __init__(self,age):
        self.dogAge = age
        
    def func1(self):
        print(self.dogAge)
        
    def func2(self,colour):
        print(f"Dog has {colour} colour")
        
obj1 = Dog(14)
obj1.func1()
obj1.func2("Brown")
```

```python
### CLASS VARIABLE & CLASS METHOD
```

```python
class School:
    school_name = "abcd"
    
School.school_name
```

```python
obj1 = School()
print(obj1.school_name)
obj1.school_name = "Greenwood High School"
print(obj1.school_name)
```

```python
class Banking:
    intRate = 0.08
    
    @classmethod
    def calculateRate(cls):
        cls.intRate = cls.intRate + (cls.intRate*0.02)
        print(cls.intRate)
        
obj2 = Banking()
Banking.intRate
obj2.calculateRate()
```
