```python
## ITERABLE

fruitsList = ["apple","mango","banana","grapes"]
print(dir(fruitsList))

for i in fruitsList:
    print(i)
```

```python
## ITERATOR

numList = [1,2,3,4,5,6,7,8]
def extractEvenNum(i):
    return i%2 == 0

abc = filter(extractEvenNum, numList)
print(abc)
print(dir(abc))

for i in abc:
    print(i)
    
```
