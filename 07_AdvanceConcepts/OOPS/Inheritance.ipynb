```python
# Single Inheritance
```

```python
class ParentClass:
    def ParentClassMethod(self):
        return "This is Parent class method"
    
class ChildClass(ParentClass):
    def ChildClassMethod(self):
        return "This is Child class method" 
    
pobj = ParentClass()
pobj.ParentClassMethod()
```

```python
cobj = ChildClass()
print(cobj.ChildClassMethod())
print(cobj.ParentClassMethod())
```

```python
# Multilevel Inheritance
```

```python
class GrandParentsClass:
    def GrandParentsMethod(self):
        return "I have 500 acres of land"
    
class ParentClass(GrandParentsClass):
    def ParentClassMethod(self):
        return "I have 50 acres of land"
    
class ChildClass(ParentClass):
    def ChildClassMethod(self):
        return "I have bike" 
    
gpobj = GrandParentsClass()
print(gpobj.GrandParentsMethod())
print(gpobj.ParentClassMethod())
```

```python
pobj2 = ParentClass()
print(pobj2.ParentClassMethod())
print(pobj2.GrandParentsMethod())
```

```python
cobj2 = ChildClass()
print(cobj2.GrandParentsMethod())
print(cobj2.ParentClassMethod())
print(cobj2.ChildClassMethod())
```

```python
# Multiple Inheritance
```

```python
class Mother:
    def motherMethod(self):
        return "I have gold"
    
class Father:
    def fatherMethod(self):
        return "I have house and money"
    
class Child(Mother, Father):
    def childMethod(self):
        return "I have books"
    

cobj3 = Child()
print(cobj3.motherMethod())
print(cobj3.fatherMethod())
print(cobj3.childMethod())

```

```python
class Mother:
    def motherMethod(self):
        return "I have gold"
    
class Son(Mother):
    def sonMethod(self):
        return "I have ps5"
    
class Daughter(Mother):
    def daughterMethod(self):
        return "I have books"
    

sobj1 = Son()
print(sobj1.motherMethod())
print(sobj1.sonMethod())

dobj1 = Daughter()
print(dobj1.motherMethod())
print(dobj1.daughterMethod())
```

```python
## REAL WORLD EXAMPLES OF INHERITANCE
```

```python
# Single

class MetroStation:
    def location(self):
        print("This is a metro station")

class PuneMetroStation(MetroStation):
    def route(self):
        print("Pune Metro")

obj1 = PuneMetroStation()
obj1.location()
obj1.route()
```

```python
# Multiple Inheritance

class Singer:
    def sing(self):
        print("Singing a song")

class Dancer:
    def dance(self):
        print("Dancing")

class Performer(Singer, Dancer):
    pass

obj2 = Performer()
obj2.sing()
obj2.dance()
```

```python
class Metro:
    def transport(self):
        print("Metro is a public transport")

class AquaLine(Metro):
    def route1(self):
        print("Aqua Line")

class PurpleLine(Metro):
    def route2(self):
        print("Purple Line")

obj4 = AquaLine()
obj4.transport()
obj4.route1()

obj5 = PurpleLine()
obj5.transport()
obj5.route2()
```

```python
# Multilevel Inheritance

class Entertainment:
    def enjoy(self):
        print("Entertainment")

class Movie(Entertainment):
    def watch(self):
        print("Watching a movie")

class ComedyMovie(Movie):
    def laugh(self):
        print("Laughing")

obj3 = ComedyMovie()
obj3.enjoy()
obj3.watch()
obj3.laugh()
```
