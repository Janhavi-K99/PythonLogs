```python
# Traditional way without using polymorphism. 
# Here we need to create different methods for different task and then call the particular method according to the requirement

class myClass:
    def sum03(self, a,b,c):
        return a+b+c
    def sum02(self, a,b):
        return a+b  
         
obj1 = myClass()
obj1.sum03(5,7,8)
obj1.sum02(6,5)
```

```python
# Polymorphism helps to call the same method and use it according to the requirement as shown in the below example

class myClass:
    def numSum(self, a=None,b=None,c=None):
        if a!=None and b!=None and c!=None :
            return a+b+c
        elif a!=None and b!=None:
            return a+b
        else:
            return "Please provide atleast 2 numbers"     
    
obj2 = myClass()
obj2.numSum(5)
    
```

```python
# METHOD OVERRIDDING
```

```python
class Animal:
    def sound(self):
        return "Animal makes sound"
    
class Dog(Animal):
    def sound(self):
        return "Dog barks"
    
class Cat(Animal):
    def sound(self):
        return "Cat meows"
        
obj3 = Dog()
print(obj3.sound())

obj4 = Cat()
print(obj4.sound())
```

```python
class Loan:
    def interest(self):
        return 0.11

class HomeLoan(Loan):
    def interest(self):
        return 0.08
    
obj5 = HomeLoan()
obj5.interest()
```

```python
# USE OF SUPER KEYWORD

class Loan:
    def interest(self):
        print("Loan interest rate: 0.11")

class HomeLoan(Loan):
    def interest(self):
        super().interest()
        print("Homeloan interest rate: 0.08")
    
obj5 = HomeLoan()
obj5.interest()
```
