TYPE CASTING IN ARRAY OR TYPE CONVERSION

```python
import numpy as np

array_2d = np.array(
    [[1,2,3,6],
     [4,5,6,8],
     [7,8,9,9]]
)

print(array_2d)
print(array_2d.dtype)
print("---------------")

array_2d_edit1 = array_2d.astype('float')
print(array_2d_edit1)
print(array_2d_edit1.dtype)
print("----------------")

array_2d_edit2 = array_2d_edit1.astype('int')
print(array_2d_edit2)
print(array_2d_edit2.dtype)
print("-----------------")
```

# ITERATION OVER ARRAY

```python
oneD_A = np.array([1,2,3,4,5,6])
for i in oneD_A:
    print(i)
```

```python
twoD_A = np.array([[1,2,3],
                  [4,5,6]])
for i in twoD_A:
    print(i)
```

```python
for i in np.nditer(twoD_A):
    print(i)
```

```python
twoD_A = np.array([[1,2,3],
                  [4,5,6]])
print(twoD_A)
print("--------------")

for i in twoD_A.T:
    print(i)
```

# STACKING

hstack

```python
arrA1 = np.array([1,2,3,4,5,6])
arrA2 = np.array([4,5,6,7,8,9])

arrA = np.hstack([arrA1,arrA2])
arrA
```

```python
np.unique(arrA)
```

vstack

```python
arrA1 = np.array([1,2,3,4,5,6])
arrA2 = np.array([4,5,6,7,8,9])
arrA3 = np.array([10,5,6,7,8,9])
arrA4 = np.array([11,5,6,7,8,90])

arrA = np.vstack([arrA1,arrA2,arrA3,arrA4])
arrA
```
