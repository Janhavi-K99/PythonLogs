```python
import pandas as pd

filepath = "D:/INTERVIEW/01 ETHANS/FOR GIT/PythonLogs/Data/Airline/Airline1.csv"
df = pd.read_csv(filepath)
df.head(2)
```

```python
import matplotlib.pyplot as plt

font = 15
plt.Figure(figsize = (7,2))
plt.bar(df['Airline'],df['Price'])
plt.xticks(rotation = 45)
plt.xlabel("Airline",fontsize = font)
plt.ylabel("Price", fontsize = font)
plt.title("Airline vs Price", fontsize = font)
plt.show()
```
