```python
# lower(): use to convert the input string into lowercase
# upper(): use to convert the input string into uppercase
# capitalise(): use to convert the first char of the input string into uppercase

txt = "Hello Everyone!!"

print(txt.lower())
print(txt.upper())
print(txt.capitalize())
```

```python
# replace(): use to replace any word or char from input with the desired word or char

txt = "Hello everyone!!"
print(txt.replace("Hello","Hi"))
```

```python
# join(): used for joinging elements from lists or tuples in order to make it a string

list1 = ["apple", "banana", "grapes"]

', '.join(list1)
```

```python
# split(): use to split a string into list or tuple
# Splits a string at a particular separater (default argument is space)

txt = "welcome to akola"
print(txt.split())

txt = "welcome,to,akola"
print(txt.split())

txt = "welcome, to, akola"
print(txt.split())


# Again you can join the result to make it a string.
```

```python
# Count(): to get the number of times particular word is repeated
# it is case-sensitive

txt = "I love apples, apples are my favorite"

txt.count("apples")
```

```python
# to get the index of the element use find or index
# find(): index number of input

txt = "I love apples, apples are my favorite fruit"

txt.find("love")
```

```python
# index(): index of input text

txt = "I love apples, apples are my favorite fruit"

txt.index("my")
```

```python
# startswith(): use to find sentences with specific entry

str_01 = "Interest of 2% given to the customer"
str_02 = "penalty of 500 charged to the customer"

str_01.startswith("Interest")
```

```python
# endswith(): use to find sentences with the specific ending

str_02.endswith("customer")
```

```python
# isnumeric(): to check if the string is numeric
txt = "5555555"
txt.isnumeric()
```

```python
# isdigit(): to check if the string is
txt = "5"
txt.isdigit()
```

```python
# isalpha(): to check if string consists of characters
txt = "sunil"
txt.isalpha()
```

```python
# isalnum(): to check if string is a combination of character & numbers
txt = "sunil0000"
txt.isalnum()
```

```python
# How to extract substring?
# Slicing is one of the option but for dynamic we need another method
# Here comes the role of string methods: isalpha() is the method we can use to differenciate the alphabates from the numerical values

# txt1 = "rajesh40"
# print(txt1[0:6])


new_txt = "123 MH30: Akola101"
new_str = ""

for i in new_txt:
    if i.isalpha() == True:
        new_str = new_str + i
        
print(new_str)
```
